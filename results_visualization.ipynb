```python
feature_types=['kncsic', 'fusionprot', 'saprot', 'protssn', 'prosst']
tasks = df['Task'].unique()
for task in tasks:
  print('\n', '#'*5, task, '#'*5)
  for feature in feature_types:
    print('\n', feature)
    if 'csic' not in feature:
      print(df[(df['Task'] == task) & (df['Feature_Type']==feature)]['Switch_F1'])
    else:
      # which rows have the string 'feature' contained in a df column
      temp_df = df[(df['Task'] == task) & (df['Feature_Type'].str.contains(feature))]
      row = temp_df.loc[temp_df['Switch_F1'].idxmax()]
      print(row['Feature_Type'], row['Switch_F1'])
```
